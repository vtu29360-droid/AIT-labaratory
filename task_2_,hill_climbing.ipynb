```python
target=10
current=2
while current!=target:
  print("Current State:",current)
  if current<target:
    current+=1
  else:
    current-=1
print("Goal Reached:",current)
```

```python
matrix=[[1,0,0],
        [0,0,1],
        [0,0,0]];
a,b=map(int,input("Start:").split())
x,y=map(int,input("Goal:").split())
path=[(a,b)]
steps=abs(x-a)+abs(y-b)
for i in range(steps):
  if a<x:
    a+=1
  elif a>x:
    a-=1
  elif b<y:
    b+=1
  elif b>y:
    b-=1
  path.append((a,b))
print("Path:",path)
print("Cost:",len(path)-1)
```
